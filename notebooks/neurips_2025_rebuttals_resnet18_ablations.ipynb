```python
import json
import numpy as np
import matplotlib.pyplot as plt

from notebook_utils import bootstrap_mean_and_ci
```

```python
def prepend_ones(arr):
    ones = np.ones((arr.shape[0], 1))
    return np.hstack((ones, arr))

def load_vision_method_soft(model, method, top_frac):
    with open(f"../scripts/_cache/soft_rates_{method}_{model}_imagenet_2_per_class_top{top_frac:.3f}.json") as f:
        return prepend_ones(np.array(json.load(f)["soft_stability_rates"]))
```

```python
resnet18_lime_0125 = load_vision_method_soft("resnet18", "lime", 0.125)
resnet18_shap_0125 = load_vision_method_soft("resnet18", "shap", 0.125)
resnet18_intg_0125 = load_vision_method_soft("resnet18", "intgrad", 0.125)
resnet18_mfaba_0125 = load_vision_method_soft("resnet18", "mfaba", 0.125)
resnet18_rand_0125 = load_vision_method_soft("resnet18", "random", 0.125)
```

```python
resnet18_lime_0250 = load_vision_method_soft("resnet18", "lime", 0.250)
resnet18_shap_0250 = load_vision_method_soft("resnet18", "shap", 0.250)
resnet18_intg_0250 = load_vision_method_soft("resnet18", "intgrad", 0.250)
resnet18_mfaba_0250 = load_vision_method_soft("resnet18", "mfaba", 0.250)
resnet18_rand_0250 = load_vision_method_soft("resnet18", "random", 0.250)
```

```python
def print_table(shaps, limes, intgs, mfabas, rands):
    shaps = shaps.mean(axis=0)
    limes = limes.mean(axis=0)
    intgs = intgs.mean(axis=0)
    mfabas = mfabas.mean(axis=0)
    rands = rands.mean(axis=0)
    
    print("| Radius | SHAP | LIME | IntGrad | MFABA | Rand |")
    print("| ------ | ---- | ---- | ------- | ----- | ---- |")
    for r in range(0,11):
        print(f"| {r} | {shaps[r]:.3f} | {limes[r]:.3f} | {intgs[r]:.3f} | {mfabas[r]:.3f} | {rands[r]:.3f} |")
```

```python
print_table(resnet18_shap_0125, resnet18_lime_0125, resnet18_intg_0125, resnet18_mfaba_0125, resnet18_rand_0125)
```

```python
print_table(resnet18_shap_0250, resnet18_lime_0250, resnet18_intg_0250, resnet18_mfaba_0250, resnet18_rand_0250)
```
