```python
import json
import numpy as np
import matplotlib.pyplot as plt
```

```python
with open("../scripts/_cache/accuracy_vs_smoothing.json") as f:
    saved = json.load(f)

lambdas = saved["lambdas"]
```

```python
def make_plot(accs, legend_title, legend_loc, fs=16):
    fig, ax = plt.subplots(figsize=(4, 3))
    ax.plot(lambdas, accs)
    plt.gca().invert_xaxis()
    
    # ax.set_xticks([1.0, 0.9, 0.8, 0.7, 0.6, 0.5, 0.4, 0.3, 0.2, 0.1])
    ax.set_xticks([1.0, 0.8, 0.6, 0.4, 0.2])
    ax.set_ylabel("Accuracy", fontsize=fs)
    ax.set_xlabel("Smoothing Parameter $\lambda$", fontsize=fs)
    
    ax.tick_params(axis="both", labelsize=fs-2)
    ax.axvline(x=1.0, color="green", linestyle="--", linewidth=2, label=f"{accs[0]:.3f}")
    ax.axvline(x=0.5, color="orange", linestyle="--", linewidth=2, label=f"{accs[5]:.3f}")
    ax.axvline(x=0.25, color="red", linestyle="--", linewidth=2, label=f"{accs[7]:.3f}")
    
    ax.legend(title=legend_title, loc=legend_loc, title_fontsize=fs, fontsize=fs)
    return fig
```

```python
fig = make_plot(saved["resnet18"], "ResNet18", "upper right")
fig.savefig("_dump/accuracy_vs_smoothing_resnet18.pdf", bbox_inches="tight")
```

```python
fig = make_plot(saved["resnet50"], "ResNet50", "upper right")
fig.savefig("_dump/accuracy_vs_smoothing_resnet50.pdf", bbox_inches="tight")
```

```python
fig = make_plot(saved["vit"], "ViT", "lower left")
fig.savefig("_dump/accuracy_vs_smoothing_vit.pdf", bbox_inches="tight")
```
